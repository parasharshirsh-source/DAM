# caping

```python
# Select the variable
col = "Ancillary_Revenue"

# Start with the original, uncapped data
before = raw[col].copy()

# Create a fresh dataset for capping
df_capped = raw.copy()

# Calculate Q1, Q3 and IQR using original data
Q1 = before.quantile(0.25)
Q3 = before.quantile(0.75)
IQR = Q3 - Q1

# Calculate the upper and lower limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Apply capping to the new dataset
df_capped[col] = before.clip(lower, upper)

# Count values capped
capped_count = ((before < lower) | (before > upper)).sum()

print("Variable:", col)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower)
print("Upper limit:", upper)
print("Values capped:", capped_count)

# Compare before and after capping
print("\nBefore capping:")
print(before.describe())

print("\nAfter capping:")
print(df_capped[col].describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Select the variable
col = "Ancillary_Revenue"

# Start with the original, uncapped data
before = raw[col].copy()

# Create a fresh dataset for capping
df_capped = raw.copy()

# Calculate Q1, Q3 and IQR using original data
Q1 = before.quantile(0.25)
Q3 = before.quantile(0.75)
IQR = Q3 - Q1

# Calculate the upper and lower limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Apply capping to the new dataset
df_capped[col] = before.clip(lower, upper)

# Print maximum values
print("Maximum before:", before.max())
print("Maximum after:", df_capped[col].max())
print("Upper limit:", upper)

# Plot before and after
plt.figure(figsize=(12, 5))

plt.plot(raw["Date"], before,
         label="Before Capping", marker="o")

plt.plot(raw["Date"], df_capped[col],
         label="After Capping", marker="x")

plt.title("Ancillary Revenue: Before vs After Capping")
plt.xlabel("Date")
plt.ylabel("Ancillary Revenue (₹)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Method 2 sub with Mean

```python
# Create a copy of the original dataset
df_mean = df.copy()

# Define numeric columns for imputation (from earlier cells)
num_cols = [
    "Passengers",
    "Ticket_Price",
    "Delay_Minutes",
    "Flight_Duration_Min",
    "Ancillary_Revenue",
    "Baggage_Complaints",
    "Fuel_Cost",
    "Customer_Rating",
    "Crew_Overtime_Hours"
]

# Display descriptive statistics BEFORE filling missing values
print("DESCRIPTIVE STATISTICS BEFORE IMPUTATION")
display(df[num_cols].describe().round(2))

# Fill missing values in numeric columns using the mean
df_mean[num_cols] = df_mean[num_cols].fillna(df_mean[num_cols].mean())

# Fill missing Route values using the most common route (mode)
most_common_route = df_mean["Route"].mode()[0]
df_mean["Route"] = df_mean["Route"].fillna(most_common_route)

# Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER MEAN AND MODE IMPUTATION")
display(df_mean[num_cols].describe().round(2))

# Compare the number of rows before and after
print("Rows before:", len(df))
print("Rows after:", len(df_mean))

print("Blanks left:", df_mean.isnull().sum().sum())
```

# Median/Imputation with Mode

```python
df_mode = df.copy()

for col in df_mode.columns:
    df_mode[col] = df_mode[col].fillna(df_mode[col].mode()[0])

print("Blanks left:", df_mode.isnull().sum().sum())
descriptive_stats = df_mode.describe().round(2)
display(df_mode.describe().round(2))
```

# Method 4 fill with median

```python
# Create a copy of the original dataset
df_median = df.copy()

# Fill missing values in numeric columns using the median
df_median[num_cols] = df_median[num_cols].fillna(
    df_median[num_cols].median()
)

# Route is text, so use the most common route
df_median["Route"] = df_median["Route"].fillna(
    df_median["Route"].mode()[0]
)

# Check remaining missing values
print("Blanks left:", df_median.isnull().sum().sum())
```

# Step - 3 Linear regression / prediction

```python
# A simple formula from our data: delivery time from distance and basket size
import pandas as pd
from sklearn.linear_model import LinearRegression

# Ensure dataset is loaded
try:
    _ = df["Distance_KM"]
except (NameError, KeyError):
    df = pd.read_excel("/content/Zepto_Orders_Data (1) (1).xlsx", sheet_name="Orders")

formula = LinearRegression().fit(
    df[["Distance_KM", "Basket_Size"]],
    df["Delivery_Time_Min"]
)

print("Each extra km adds about", round(formula.coef_[0], 1), "minutes")
print("Each extra item adds about", round(formula.coef_[1], 2), "minutes")
```

Step -4 Developing what if case

```python
# Fill every what-if case with the expected delivery time
cases = pd.DataFrame({
    "Distance_KM": D.ravel(),
    "Basket_Size": B.ravel()
})

T = formula.predict(cases).reshape(D.shape)
```

# How fast can a store delivers  for a given basket and still keep the 10 min promiss

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Recreate the meshgrid what-if cases
dist_values   = np.linspace(0.3, 4, 40)
basket_values = np.linspace(1, 12, 40)
D, B = np.meshgrid(dist_values, basket_values)

# Re-predict using our regression model
cases = pd.DataFrame({
    "Distance_KM": D.ravel(),
    "Basket_Size": B.ravel()
})
T = formula.predict(cases).reshape(D.shape)

# Create the contour plot
plt.contourf(D, B, T, levels=10, cmap="Purples")
plt.colorbar(label="Expected delivery time (minutes)")

lines = plt.contour(D, B, T, levels=[8, 10, 12, 15], colors="red")
plt.clabel(lines, fmt="%d min")

plt.title("Distance and basket size that keep the 10-minute promise")
plt.xlabel("Distance from store (km)")
plt.ylabel("Basket size (items)")
plt.show()
```

# lecture - 11 / how did the sales build up week by week in each state

```python
import pandas as pd
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Ensure the dataset is loaded and helper columns are defined
try:
    _ = df["Week"]
except (NameError, KeyError):
    df = pd.read_excel("/content/Zepto_Orders_Data (1) (1).xlsx", sheet_name="Orders")
    df["Order_Date"] = pd.to_datetime(df["Order_Date"])
    df["Week"] = df["Order_Date"].dt.isocalendar().week.astype(int)

# Running total of sales, week by week, for each state
running = df.pivot_table(
    index='Week',
    columns='State',
    values='Order_Value',
    aggfunc='sum'
).fillna(0).cumsum()

running = running[running.iloc[-1].sort_values().index]  # smallest to largest
```

```python
groups = {
    "Customer Experience": ["Taste_Score", "Service_Score", "Cleanliness_Score", "Ambience_Score", "Staff_Friendliness_Score"],
    "Digital Engagement": ["App_Orders_Share_Pct", "Loyalty_Members_K", "App_Rating", "Social_Followers_K"],
    "Marketing Intensity": ["Social_Media_Spend_K", "Outdoor_Ad_Spend_K", "Influencer_Spend_K", "Promo_Offers_Count"],
    "Store Capacity": ["Store_Area_SqFt", "Seats", "Staff_Count"],
}

for name, cols in groups.items():
    c = df[cols].corr().values
    average = (c.sum() - len(cols)) / (len(cols) * (len(cols) - 1))  # average correlation inside the group
    print(f"{name}: {len(cols)} variables, average correlation {average:.2f}")
```

PC customer experience index

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

cols = groups["Customer Experience"]

# Step A: put every variable on the same scale
scaled = StandardScaler().fit_transform(df[cols])

# Step B: run PCA
pca = PCA()
scores = pca.fit_transform(scaled)

# Step C: how much does each component explain?
names = [f"PC{i+1}" for i in range(len(cols))]
print(pd.Series(pca.explained_variance_ratio_, index=names).round(2))
```

```python
# Step D: the loadings show how much each variable counts in PC1

sign = 1 if pca.components_[0].sum() > 0 else -1  # make sure a higher score means better

print(pd.Series(pca.components_[0] * sign, index=cols).round(2))
```

```python
# Step E: turn PC1 into an index from 0 to 100

raw_index = scores[:, 0] * sign

df["Experience_Index"] = (raw_index - raw_index.min()) / (raw_index.max() - raw_index.min()) * 100

variance_explained = {"Experience_Index": round(pca.explained_variance_ratio_[0] * 100)}

print(df[["Store_ID", "Experience_Index"]].sort_values("Experience_Index").head(5).round(1))
df.head()
```

pca 2 digital engagement index

```python
def make_index(columns, index_name):
    scaled = StandardScaler().fit_transform(df[columns])
    pca = PCA().fit(scaled)
    scores = pca.transform(scaled)[:, 0]

    sign = 1 if pca.components_[0].sum() > 0 else -1
    scores = scores * sign

    df[index_name] = (scores - scores.min()) / (scores.max() - scores.min()) * 100
    variance_explained[index_name] = round(pca.explained_variance_ratio_[0] * 100)

    print("Variance explained by PC1:", variance_explained[index_name], "%")
    print(pd.Series(pca.components_[0] * sign, index=columns).round(2))


make_index(groups["Digital Engagement"], "Digital_Index")
```
